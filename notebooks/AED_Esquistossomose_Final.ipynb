# Tratamento de Valores Ausentes

### Conversão de Código UF IBGE para a sigla UF

https://www.ibge.gov.br/explica/codigos-dos-municipios.php

```python
# @title
#Converte Código UF IBGE para a sigla UF
mapa_uf = {
    '11':'RO', '12':'AC', '13':'AM', '14':'RR', '15':'PA', '16':'AP', '17':'TO',
    '21':'MA', '22':'PI', '23':'CE', '24':'RN', '25':'PB', '26':'PE', '27':'AL', '28':'SE', '29':'BA',
    '31':'MG', '32':'ES', '33':'RJ', '35':'SP',
    '41':'PR', '42':'SC', '43':'RS',
    '50':'MS', '51':'MT', '52':'GO', '53':'DF'
}


df['SG_UF'] = df['SG_UF'].astype(str)  # garante string
df['SG_UF'] = df['SG_UF'].map(mapa_uf)
df['COUFINF'] = (
    df['COUFINF']
    .astype(str)
    .str.strip()
    .str.replace('.0', '', regex=False)
    .map(mapa_uf)
)

```

##1 Análise sobre Notificação

Aqui, queremos verificar quais estados do Brasil possuem maior número de notificação de Esquistossomosse

```python
# @title
import matplotlib.pyplot as plt

# Estado da notificacao (codigo IBGE -> sigla)
uf_not = (
    df['SG_UF_NOT']
    .astype(str)
    .str[:2]
    .map(mapa_uf)
)

casos_uf = uf_not.value_counts().sort_values(ascending=False)

# TOP 10
casos_uf = casos_uf.head(10)

# porcentagem
incidencia_uf = (casos_uf / casos_uf.sum() * 100).round(2)

# --- CORES EM DEGRADÊ ---
cmap = plt.cm.viridis  # opções: plasma, Blues, coolwarm
cores = cmap(np.linspace(0.3, 0.9, len(casos_uf)))

# Grafico
fig, ax = plt.subplots(figsize=(10, 5))
bars = ax.bar(casos_uf.index, casos_uf.values, color=cores)

# porcentagem
for bar, pct in zip(bars, incidencia_uf.values):
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f'{pct}%',
        ha='center',
        va='bottom',
        fontsize=9
    )

ax.set_title('Top 10 UFs por Incidencia de Notificacao')
ax.set_xlabel('UF')
ax.set_ylabel('Casos')
ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# @title

df['DT_NOTIFIC'] = pd.to_datetime(df['DT_NOTIFIC'], errors='coerce')
df['NU_ANO'] = pd.to_numeric(df['NU_ANO'], errors='coerce')


def gerar_series(dataframe):

    sem_not_str = (
        dataframe['SEM_NOT']
        .astype(str)
        .str.extract(r'(\d{4,6})', expand=False)
        .str.zfill(6)
    )

    ano_sem_not = pd.to_numeric(sem_not_str.str[:4], errors='coerce')
    semana_not = pd.to_numeric(sem_not_str.str[-2:], errors='coerce')

    # --- série temporal ---
    ano_notificacao = (
        dataframe['DT_NOTIFIC'].dt.year
        .fillna(dataframe['NU_ANO'])
        .fillna(ano_sem_not)
    )

    ano_notificacao = ano_notificacao.where((ano_notificacao >= 1900) & (ano_notificacao <= 2100))
    casos_ano = ano_notificacao.dropna().astype(int).value_counts().sort_index()

    # --- sazonalidade ---
    semanas_validas = semana_not.where((semana_not >= 1) & (semana_not <= 53))
    casos_semana = semanas_validas.dropna().astype(int).value_counts().sort_index()

    return casos_ano, casos_semana


# Brasil
casos_ano_br, casos_sem_br = gerar_series(df)

# MG
df_mg = df[df['SG_UF'].astype(str).str[:2] == 'MG']
casos_ano_mg, casos_sem_mg = gerar_series(df_mg)

# CORES (viridis)
cmap = plt.cm.viridis

# PLOT 2x2
fig, axs = plt.subplots(2, 2, figsize=(16, 10))

# --- (1) Série temporal Brasil ---
axs[0, 0].plot(
    casos_ano_br.index, casos_ano_br.values,
    marker='o', linewidth=2, color=cmap(0.6)
)
axs[0, 0].set_title('Brasil - Casos por Ano')
axs[0, 0].set_xlabel('Ano')
axs[0, 0].set_ylabel('Casos')
axs[0, 0].grid(alpha=0.3)

# --- (2) Série temporal MG ---
axs[0, 1].plot(
    casos_ano_mg.index, casos_ano_mg.values,
    marker='o', linewidth=2, color=cmap(0.85)
)
axs[0, 1].set_title('MG - Casos por Ano')
axs[0, 1].set_xlabel('Ano')
axs[0, 1].set_ylabel('Casos')
axs[0, 1].grid(alpha=0.3)

# --- (3) Sazonalidade Brasil ---
cores_br = cmap(np.linspace(0.3, 0.9, len(casos_sem_br)))

axs[1, 0].bar(
    casos_sem_br.index, casos_sem_br.values,
    color=cores_br
)
axs[1, 0].set_title('Brasil - Sazonalidade')
axs[1, 0].set_xlabel('Semana')
axs[1, 0].set_ylabel('Casos')
axs[1, 0].set_xticks(range(1, 54, 4))
axs[1, 0].grid(axis='y', alpha=0.3)

# --- (4) Sazonalidade MG ---
cores_mg = cmap(np.linspace(0.3, 0.9, len(casos_sem_mg)))

axs[1, 1].bar(
    casos_sem_mg.index, casos_sem_mg.values,
    color=cores_mg
)
axs[1, 1].set_title('MG - Sazonalidade')
axs[1, 1].set_xlabel('Semana')
axs[1, 1].set_ylabel('Casos')
axs[1, 1].set_xticks(range(1, 54, 4))
axs[1, 1].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

Observação sobre a semana 53

Nem todos os anos terminam na semana epidemiológica 53; a maioria termina na semana 52. Por isso a semana 53 apresenta um número significativamente menor de notificações, o que pode distorcer análises de sazonalidade se não for considerado.

```python
# @title

# =========================
# PREPARAÇÃO
# =========================
nu_idade_str = (
    df['NU_IDADE_N']
    .astype(str)
    .str.extract(r'(\d+)', expand=False)
    .str.zfill(4)
)

df['IDADE_DERIVADA'] = pd.to_numeric(nu_idade_str.str[-2:], errors='coerce')

idade = df['IDADE_DERIVADA']
sexo = df['CS_SEXO'].astype(str).str.strip()
raca = pd.to_numeric(df['CS_RACA'], errors='coerce')
escolaridade = pd.to_numeric(df['CS_ESCOL_N'], errors='coerce')

# =========================
# LIMPEZA (REMOVER IGNORADOS)
# =========================
sexo = sexo[~sexo.isin(['I', 'Ignorado', 'nan', ''])]
raca = raca[~raca.isin([9])]
escolaridade = escolaridade[~escolaridade.isin([9, 10])]

# =========================
# FAIXAS E MAPAS
# =========================
faixa_idade = pd.cut(
    idade,
    bins=[0, 5, 12, 18, 30, 50, 70, 120],
    labels=['0-5', '6-12', '13-18', '19-30', '31-50', '51-70', '71+']
)

map_raca = {
    1: 'Branca', 2: 'Preta', 3: 'Amarela', 4: 'Parda', 5: 'Indigena'
}
raca_label = raca.map(map_raca).dropna()

map_escol = {
    0: 'Analfabeto', 1: '1a-4a incompl.', 2: '4a compl.', 3: '5a-8a incompl.',
    4: 'Fundamental compl.', 5: 'Medio incompl.', 6: 'Medio compl.',
    7: 'Superior incompl.', 8: 'Superior compl.'
}
escol_label = escolaridade.map(map_escol).dropna()

# =========================
# DISTRIBUIÇÕES (%)
# =========================
def distribuicao_percentual(serie):
    dist = serie.value_counts()
    pct = (dist / dist.sum() * 100).round(2)
    return dist, pct

sexo_dist, sexo_pct = distribuicao_percentual(sexo)
idade_dist, idade_pct = distribuicao_percentual(faixa_idade)
raca_dist, raca_pct = distribuicao_percentual(raca_label)
escol_dist, escol_pct = distribuicao_percentual(escol_label)

# =========================
# CORES (PALETA BONITA)
# =========================
cmap = plt.cm.viridis  # pode trocar: viridis, coolwarm, inferno

# =========================
# PLOT
# =========================
fig, axs = plt.subplots(2, 2, figsize=(14, 9))

def plot_bar(ax, dist, pct, titulo):
    cores = cmap(np.linspace(0.3, 0.9, len(dist)))
    bars = ax.bar(dist.index.astype(str), dist.values, color=cores)

    # porcentagem
    for bar, p in zip(bars, pct.values):
        ax.text(
            bar.get_x() + bar.get_width()/2,
            bar.get_height(),
            f'{p}%',
            ha='center',
            va='bottom',
            fontsize=9
        )

    ax.set_title(titulo)
    ax.grid(axis='y', alpha=0.3)

# --- gráficos ---
plot_bar(axs[0, 0], sexo_dist, sexo_pct, 'Casos por Sexo')
plot_bar(axs[0, 1], idade_dist, idade_pct, 'Casos por Faixa Etaria')
plot_bar(axs[1, 0], raca_dist, raca_pct, 'Casos por Raca/Cor')
plot_bar(axs[1, 1], escol_dist, escol_pct, 'Casos por Escolaridade')

# ajustes visuais
axs[1, 0].tick_params(axis='x', rotation=25)
axs[1, 1].tick_params(axis='x', rotation=30)

plt.tight_layout()
plt.show()

# =========================
# CRUZAMENTO
# =========================
cruzamento = pd.crosstab(sexo, raca_label, normalize='all') * 100

print('\nTop combinacoes Sexo x Raca (% do total):')
print(cruzamento.stack().sort_values(ascending=False).head(10).round(2))

# =========================
# NOVA ANÁLISE: Comparação Raça/Cor com População Brasileira
# =========================

# Dados aproximados da população brasileira (IBGE Censo 2022, arredondado)
populacao_br_raca = pd.Series({
    'Branca': 43.5,
    'Parda': 45.3,
    'Preta': 10.6,
    'Amarela': 0.4,
    'Indigena': 0.8
}).sort_index()

# Reordena a distribuição da amostra para corresponder à ordem da população brasileira
raca_pct_sorted = raca_pct.reindex(populacao_br_raca.index, fill_value=0).sort_index()

# DataFrame para comparação
comparacao_raca = pd.DataFrame({
    'Amostra (%)': raca_pct_sorted,
    'População Brasileira (%)': populacao_br_raca
})

print('\nComparação da Distribuição de Raça/Cor:')
print(comparacao_raca)

# Plot da comparação
fig_comp, ax_comp = plt.subplots(figsize=(10, 6))

comparacao_raca.plot(kind='bar', ax=ax_comp, color=[cmap(0.4), cmap(0.7)])

ax_comp.set_title('Distribuição de Raça/Cor: Amostra vs. População Brasileira')
ax_comp.set_ylabel('Percentual (%)')
ax_comp.set_xlabel('Raça/Cor')
ax_comp.tick_params(axis='x', labelrotation=45)
plt.setp(ax_comp.get_xticklabels(), ha='right')
ax_comp.grid(axis='y', alpha=0.3)

# Adicionar porcentagens nas barras
for container in ax_comp.containers:
    for i, bar in enumerate(container):
        value = bar.get_height()
        ax_comp.text(
            bar.get_x() + bar.get_width() / 2,
            value + 0.5, # Posição vertical um pouco acima da barra
            f'{value:.1f}%',
            ha='center',
            va='bottom',
            fontsize=8
        )

plt.tight_layout()
plt.show()
```

##4 Análise de Dados Clínicos e Investigação

Nesta seção, analisaremos atributos das instâncias que contém dados clínicos e de ocupação do paciente:

* `DT_SIN_PRI`: Data dos primeiros sintomas. Formato em AAAA-MM-DD (Exemplo: 2007-11-01 para 1 de novembro de 2007)

* `SEM_PRI`: Semana epidemiológica dos sintomas. Formato float em AAAASS.F (Exemplo 200744.0 para semana 44 de 2007)

* `DT_INVEST`: Data da investigação. Formato Float em AAAAMMDD.F (Exemplo: 20071101.0 para 1 de novembro de 2007)


Tarefas:

- Verificação de valores ausentes nesses atributos, para garantir que temos dados suficientes para análise.
- Análise de tempo entre início dos sintomas e investigação, para avaliar a eficiência do sistema de saúde.

```python
# @title
# Tarefa 1: verificar valores ausentes/invalidos nos atributos clinicos e de investigacao
import matplotlib.pyplot as plt

cols_clinicas = ['DT_SIN_PRI', 'SEM_PRI', 'DT_INVEST']

# Copia para evitar alterar formatos originais diretamente
tmp = df[cols_clinicas].copy()
tmp = tmp.replace(['', ' '], np.nan)

# DT_SIN_PRI: formato esperado AAAA-MM-DD
dt_sin_pri_dt = pd.to_datetime(tmp['DT_SIN_PRI'], format='%Y-%m-%d', errors='coerce')

# -------------------- Diagnostico detalhado de SEM_PRI --------------------
# Formato esperado: AAAASS.F (ex.: 200744.0), onde semana deve estar entre 01 e 53
sem_raw = tmp['SEM_PRI']
sem_num = pd.to_numeric(sem_raw, errors='coerce')
sem_int = pd.Series(pd.NA, index=tmp.index, dtype='Int64')
sem_int.loc[sem_num.notna()] = sem_num.loc[sem_num.notna()].round().astype('Int64')
sem_str = sem_int.astype('string')

# Regras de qualidade
sem_missing = sem_raw.isna()
sem_non_numeric = sem_raw.notna() & sem_num.isna()
sem_len = sem_str.str.len()
sem_short = sem_num.notna() & (sem_len < 6)
sem_long = sem_num.notna() & (sem_len > 6)

# Caso estritamente valido: 6 digitos AAAASS
sem_6 = sem_str.where(sem_len == 6)
sem_ano_6 = pd.to_numeric(sem_6.str[:4], errors='coerce')
sem_semana_6 = pd.to_numeric(sem_6.str[4:6], errors='coerce')
sem_valido_6 = (
    sem_6.notna()
    & sem_ano_6.between(1900, 2100, inclusive='both')
    & sem_semana_6.between(1, 53, inclusive='both')
)
sem_week_out_6 = sem_6.notna() & ~sem_semana_6.between(1, 53, inclusive='both')
sem_year_out_6 = sem_6.notna() & ~sem_ano_6.between(1900, 2100, inclusive='both')

# Heuristica para valores curtos (3-4 digitos): interpretar como YYWW
sem_short_yyww = sem_str.where(sem_len.isin([3, 4])).str.zfill(4)
sem_short_yy = pd.to_numeric(sem_short_yyww.str[:2], errors='coerce')
sem_short_ww = pd.to_numeric(sem_short_yyww.str[2:4], errors='coerce')
sem_short_recuperavel = (
    sem_short_yyww.notna()
    & sem_short_ww.between(1, 53, inclusive='both')
)

# Conversao principal de SEM_PRI usada na analise (apenas casos estritamente validos)
sem_pri_ano = sem_ano_6
sem_pri_semana = sem_semana_6
sem_pri_valido = sem_valido_6

# DT_INVEST: formato esperado AAAAMMDD.F (ex.: 20071101.0)
dt_invest_num = pd.to_numeric(tmp['DT_INVEST'], errors='coerce')
dt_invest_int = pd.Series(pd.NA, index=tmp.index, dtype='Int64')
dt_invest_int.loc[dt_invest_num.notna()] = dt_invest_num.loc[dt_invest_num.notna()].round().astype('Int64')
dt_invest_str = dt_invest_int.astype('string').str.extract(r'(\d{8})', expand=False)
dt_invest_dt = pd.to_datetime(dt_invest_str, format='%Y%m%d', errors='coerce')

# Salva colunas tratadas para a proxima tarefa
df['DT_SIN_PRI_DT'] = dt_sin_pri_dt
df['SEM_PRI_ANO'] = sem_pri_ano
df['SEM_PRI_SEMANA'] = sem_pri_semana.where(sem_pri_valido)
df['DT_INVEST_DT'] = dt_invest_dt

# Resumo de ausentes + invalidos apos padronizacao
validos = pd.Series({
    'DT_SIN_PRI': dt_sin_pri_dt.notna().sum(),
    'SEM_PRI': sem_pri_valido.sum(),
    'DT_INVEST': dt_invest_dt.notna().sum(),
})

total = len(df)
aus_invalidos = total - validos
pct_aus_invalidos = (aus_invalidos / total * 100).round(2)

resumo_clin = pd.DataFrame({
    'Validos': validos,
    'Ausentes_ou_Invalidos': aus_invalidos,
    'Percentual_Aus_Inv (%)': pct_aus_invalidos,
}).sort_values('Percentual_Aus_Inv (%)', ascending=False)

print('Resumo de completude apos validacao de formato:')
print(resumo_clin)

fig, ax = plt.subplots(figsize=(8, 4))
ax.bar(resumo_clin.index, resumo_clin['Percentual_Aus_Inv (%)'])
ax.set_title('Ausentes/Invalidos - Dados Clinicos e Investigacao')
ax.set_ylabel('Percentual (%)')
ax.set_xlabel('Atributo')
ax.set_ylim(0, max(5, resumo_clin['Percentual_Aus_Inv (%)'].max() * 1.2))
for i, v in enumerate(resumo_clin['Percentual_Aus_Inv (%)'].values):
    ax.text(i, v + 0.05, f'{v:.2f}%', ha='center', fontsize=9)
ax.grid(alpha=0.3)
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()

# Relatorio especifico de inconsistencias de SEM_PRI
sem_issues = pd.Series({
    'Ausente': sem_missing.sum(),
    'Nao numerico': sem_non_numeric.sum(),
    'Curto (<6 digitos)': sem_short.sum(),
    'Longo (>6 digitos)': sem_long.sum(),
    'Ano fora da faixa (6 digitos)': sem_year_out_6.sum(),
    'Semana fora de 01-53 (6 digitos)': sem_week_out_6.sum(),
    'Validos estritos (AAAASS)': sem_pri_valido.sum(),
    'Curtos recuperaveis (heuristica YYWW)': sem_short_recuperavel.sum(),
}, dtype='int64')

sem_issues_pct = (sem_issues / total * 100).round(2)
diagnostico_sem = pd.DataFrame({
    'Quantidade': sem_issues,
    'Percentual (%)': sem_issues_pct,
})

print('\nDiagnostico de inconsistencias em SEM_PRI:')
print(diagnostico_sem)

# Mostra os valores estranhos mais frequentes
mask_estranho = sem_non_numeric | sem_short | sem_long | sem_week_out_6 | sem_year_out_6
valores_estranhos = sem_raw[mask_estranho].astype(str).value_counts().head(20)

print('\nTop 20 valores estranhos de SEM_PRI:')
if len(valores_estranhos) > 0:
    print(valores_estranhos)
else:
    print('Nenhum valor estranho encontrado.')

fig, ax = plt.subplots(figsize=(10, 4))
ax.bar(diagnostico_sem.index, diagnostico_sem['Percentual (%)'])
ax.set_title('SEM_PRI - Tipos de Inconsistencia')
ax.set_ylabel('Percentual (%)')
ax.set_xlabel('Categoria')
ax.grid(alpha=0.3)
plt.xticks(rotation=35, ha='right')
plt.tight_layout()
#plt.show()
```

```python
# @title
# Tarefa 2: analise do tempo entre inicio dos sintomas e investigacao
import matplotlib.pyplot as plt

# Garante que as colunas tratadas existem (caso a celula anterior nao tenha sido executada)
if 'DT_SIN_PRI_DT' not in df.columns:
    df['DT_SIN_PRI_DT'] = pd.to_datetime(df['DT_SIN_PRI'], format='%Y-%m-%d', errors='coerce')

if 'DT_INVEST_DT' not in df.columns:
    dt_invest_num = pd.to_numeric(df['DT_INVEST'], errors='coerce')
    dt_invest_int = pd.Series(pd.NA, index=df.index, dtype='Int64')
    dt_invest_int.loc[dt_invest_num.notna()] = dt_invest_num.loc[dt_invest_num.notna()].round().astype('Int64')
    dt_invest_str = dt_invest_int.astype('string').str.extract(r'(\d{8})', expand=False)
    df['DT_INVEST_DT'] = pd.to_datetime(dt_invest_str, format='%Y%m%d', errors='coerce')

if 'SEM_PRI_SEMANA' not in df.columns or 'SEM_PRI_ANO' not in df.columns:
    sem_num = pd.to_numeric(df['SEM_PRI'], errors='coerce')
    sem_int = pd.Series(pd.NA, index=df.index, dtype='Int64')
    sem_int.loc[sem_num.notna()] = sem_num.loc[sem_num.notna()].round().astype('Int64')
    sem_str = sem_int.astype('string')
    sem_len = sem_str.str.len()
    sem_6 = sem_str.where(sem_len == 6)
    sem_ano = pd.to_numeric(sem_6.str[:4], errors='coerce')
    sem_semana = pd.to_numeric(sem_6.str[4:6], errors='coerce')
    sem_valido = (
        sem_6.notna()
        & sem_ano.between(1900, 2100, inclusive='both')
        & sem_semana.between(1, 53, inclusive='both')
    )
    df['SEM_PRI_ANO'] = sem_ano
    df['SEM_PRI_SEMANA'] = sem_semana.where(sem_valido)

# Usa apenas dados SEM inconsistencias nas variaveis da tarefa
mask_consistente_base = (
    df['DT_SIN_PRI_DT'].notna()
    & df['DT_INVEST_DT'].notna()
    & df['SEM_PRI_ANO'].notna()
    & df['SEM_PRI_SEMANA'].notna()
)

# Checagem adicional leve: ano em SEM_PRI deve estar no maximo a 1 ano do ano de DT_SIN_PRI
ano_sintoma = df['DT_SIN_PRI_DT'].dt.year
mask_ano_compativel = (df['SEM_PRI_ANO'] - ano_sintoma).abs() <= 1

# Diferença em dias: investigacao - inicio dos sintomas
tempo_invest_dias = (df['DT_INVEST_DT'] - df['DT_SIN_PRI_DT']).dt.days

# Mantem somente consistentes e com tempo nao negativo
mask_tempo_nao_negativo = tempo_invest_dias >= 0

# Exclui atrasos extremos maiores que 365 dias para reduzir distorcao de outliers
mask_sem_atraso_extremo = tempo_invest_dias <= 365

mask_consistente = (
    mask_consistente_base
    & mask_ano_compativel
    & mask_tempo_nao_negativo
    & mask_sem_atraso_extremo
)

tempo_analise = tempo_invest_dias[mask_consistente].dropna()

# Relatorio de filtragem (para Brasil)
total_registros = len(df)
base_consistente = mask_consistente_base.sum()
ano_incompativel = (mask_consistente_base & ~mask_ano_compativel).sum()
tempo_negativo = (mask_consistente_base & mask_ano_compativel & ~mask_tempo_nao_negativo).sum()
atraso_extremo = (
    mask_consistente_base
    & mask_ano_compativel
    & mask_tempo_nao_negativo
    & ~mask_sem_atraso_extremo
).sum()
usados_final = len(tempo_analise)

print(f'Total de registros: {total_registros}')
print(f'Com variaveis validas (DT_SIN_PRI, DT_INVEST, SEM_PRI): {base_consistente}')
print(f'Excluidos por ano SEM_PRI incompativel com DT_SIN_PRI: {ano_incompativel}')
print(f'Excluidos por tempo negativo: {tempo_negativo}')
print(f'Excluidos por atraso extremo (>365 dias): {atraso_extremo}')
print(f'Registros finais sem inconsistencias usados na analise: {usados_final}')

if len(tempo_analise) > 0:
    media = tempo_analise.mean()
    mediana = tempo_analise.median()
    p90 = tempo_analise.quantile(0.90)
    p95 = tempo_analise.quantile(0.95)

    ate_7 = (tempo_analise <= 7).mean() * 100
    ate_14 = (tempo_analise <= 14).mean() * 100
    ate_30 = (tempo_analise <= 30).mean() * 100

    print('\nIndicadores de oportunidade da investigacao (somente dados consistentes e sem outliers extremos):')
    print(f'- Media: {media:.2f} dias')
    print(f'- Mediana: {mediana:.2f} dias')
    print(f'- P90: {p90:.2f} dias')
    print(f'- P95: {p95:.2f} dias')
    print(f'- Investigados em ate 7 dias: {ate_7:.2f}%')
    print(f'- Investigados em ate 14 dias: {ate_14:.2f}%')
    print(f'- Investigados em ate 30 dias: {ate_30:.2f}%')

    # Tendencia temporal (mediana por ano de inicio dos sintomas) para Brasil
    serie_tempo = pd.DataFrame({
        'ano_sintoma': ano_sintoma,
        'tempo_dias': tempo_invest_dias,
    })
    serie_tempo = serie_tempo[mask_consistente].dropna()
    mediana_por_ano = serie_tempo.groupby('ano_sintoma')['tempo_dias'].median().sort_index()

    # --- INICIO: Analise para MG ---
    df_mg = df[df['SG_UF'] == 'MG'].copy() # Filtra dados para MG

    mask_consistente_base_mg = (
        df_mg['DT_SIN_PRI_DT'].notna()
        & df_mg['DT_INVEST_DT'].notna()
        & df_mg['SEM_PRI_ANO'].notna()
        & df_mg['SEM_PRI_SEMANA'].notna()
    )
    ano_sintoma_mg = df_mg['DT_SIN_PRI_DT'].dt.year
    mask_ano_compativel_mg = (df_mg['SEM_PRI_ANO'] - ano_sintoma_mg).abs() <= 1
    tempo_invest_dias_mg = (df_mg['DT_INVEST_DT'] - df_mg['DT_SIN_PRI_DT']).dt.days
    mask_tempo_nao_negativo_mg = tempo_invest_dias_mg >= 0
    mask_sem_atraso_extremo_mg = tempo_invest_dias_mg <= 365

    mask_consistente_mg = (
        mask_consistente_base_mg
        & mask_ano_compativel_mg
        & mask_tempo_nao_negativo_mg
        & mask_sem_atraso_extremo_mg
    )
    tempo_analise_mg = tempo_invest_dias_mg[mask_consistente_mg].dropna()

    serie_tempo_mg = pd.DataFrame({
        'ano_sintoma': ano_sintoma_mg,
        'tempo_dias': tempo_invest_dias_mg,
    })
    serie_tempo_mg = serie_tempo_mg[mask_consistente_mg].dropna()
    mediana_por_ano_mg = serie_tempo_mg.groupby('ano_sintoma')['tempo_dias'].median().sort_index()
    # --- FIM: Analise para MG ---

    # --- Plotagem 2x2 para Tempo de Investigação (Brasil vs MG) ---
    fig, axs = plt.subplots(2, 2, figsize=(14, 10))
    cmap_hist = plt.cm.viridis # Colormap para os histogramas e linhas

    # Histograma Brasil (Superior Esquerda)
    limite_hist = max(1, float(p95)) # Usa p95 do Brasil
    dados_hist = tempo_analise[tempo_analise <= limite_hist]
    axs[0, 0].hist(dados_hist, bins=30, color=cmap_hist(0.4))
    axs[0, 0].axvline(mediana, linestyle='--', color='red', label=f'Mediana BR: {mediana:.1f}d')
    axs[0, 0].set_title('BR - Tempo para Investigação (dias)')
    axs[0, 0].set_xlabel('Dias (até P95 BR)')
    axs[0, 0].set_ylabel('Frequência')
    axs[0, 0].legend()
    axs[0, 0].grid(alpha=0.3)

    # Histograma MG (Superior Direita)
    if len(tempo_analise_mg) > 0:
        mediana_mg = tempo_analise_mg.median()
        p95_mg = tempo_analise_mg.quantile(0.95)
        limite_hist_mg = max(1, float(p95_mg))
        dados_hist_mg = tempo_analise_mg[tempo_analise_mg <= limite_hist_mg]
        axs[0, 1].hist(dados_hist_mg, bins=30, color=cmap_hist(0.7))
        axs[0, 1].axvline(mediana_mg, linestyle='--', color='red', label=f'Mediana MG: {mediana_mg:.1f}d')
        axs[0, 1].set_title('MG - Tempo para Investigação (dias)')
        axs[0, 1].set_xlabel('Dias (até P95 MG)')
        axs[0, 1].set_ylabel('Frequência')
        axs[0, 1].legend()
        axs[0, 1].grid(alpha=0.3)
    else:
        axs[0, 1].set_title('MG - Tempo para Investigação (dias)')
        axs[0, 1].text(0.5, 0.5, 'Dados insuficientes para MG', horizontalalignment='center', verticalalignment='center', transform=axs[0, 1].transAxes)

    # Série Temporal Brasil (Inferior Esquerda)
    anos_plot = mediana_por_ano.index.astype(int)
    axs[1, 0].plot(anos_plot, mediana_por_ano.values, marker='o', color=cmap_hist(0.4), label='Brasil')
    axs[1, 0].set_title('BR - Mediana Tempo de Busca por Atendimento')
    axs[1, 0].set_xlabel('Ano')
    axs[1, 0].set_ylabel('Mediana (dias)')
    axs[1, 0].set_xticks(anos_plot)
    axs[1, 0].set_xticklabels(anos_plot.astype(str), rotation=75, ha='right')
    axs[1, 0].legend()
    axs[1, 0].grid(alpha=0.3)

    # Série Temporal MG (Inferior Direita)
    if len(mediana_por_ano_mg) > 0:
        anos_plot_mg = mediana_por_ano_mg.index.astype(int)
        axs[1, 1].plot(anos_plot_mg, mediana_por_ano_mg.values, marker='o', color=cmap_hist(0.7), label='MG')
        axs[1, 1].set_title('MG - Mediana Tempo de Busca por Atendimento')
        axs[1, 1].set_xlabel('Ano')
        axs[1, 1].set_ylabel('Mediana (dias)')
        axs[1, 1].set_xticks(anos_plot_mg)
        axs[1, 1].set_xticklabels(anos_plot_mg.astype(str), rotation=75, ha='right')
        axs[1, 1].legend()
        axs[1, 1].grid(alpha=0.3)
    else:
        axs[1, 1].set_title('MG - Mediana Tempo de Busca por Atendimento')
        axs[1, 1].text(0.5, 0.5, 'Dados insuficientes para MG', horizontalalignment='center', verticalalignment='center', transform=axs[1, 1].transAxes)

    plt.tight_layout()
    plt.show()
else:
    print('Nao há registros suficientes sem inconsistências para estimar o tempo de investigação.')

# Existing plots from earlier in the cell (applying viridis colormap)
cmap_bar = plt.cm.viridis
```

```python
# @title
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy.stats import chi2_contingency

cmap = plt.cm.viridis

# =========================
# PREPARAÇÃO DOS DADOS
# =========================

# Óbito (ajuste se necessário)
df['OBITO_BIN'] = df['EVOLUCAO'].apply(lambda x: 1 if x in [3,4] else 0)

# Idade numérica
df['IDADE_DERIVADA'] = pd.to_numeric(df['IDADE_DERIVADA'], errors='coerce')

# Faixa etária
df['FAIXA_IDADE'] = pd.cut(
    df['IDADE_DERIVADA'],
    bins=[0, 18, 40, 60, 80, 120],
    labels=['0-18','19-40','41-60','61-80','80+']
)

# Garantir variáveis numéricas
df['TRATAM'] = pd.to_numeric(df['TRATAM'], errors='coerce')
df['FORMA'] = pd.to_numeric(df['FORMA'], errors='coerce')

# Remover sexo indeterminado
df_filtrado = df[df['CS_SEXO'] != 'I'].copy()

# =========================
# FUNÇÃO DE ANÁLISE
# =========================

def analisar_associacao(dataframe, var):

    tabela = pd.crosstab(dataframe[var], dataframe['OBITO_BIN'])

    # remover categorias vazias
    tabela = tabela[tabela.sum(axis=1) > 0]

    # porcentagem de óbito
    if 1 in tabela.columns:
        perc = (tabela[1] / tabela.sum(axis=1)) * 100
    else:
        perc = pd.Series(0, index=tabela.index)

    perc = perc.sort_values(ascending=False)

    # teste estatístico seguro
    try:
        if tabela.shape[0] > 1 and tabela.shape[1] > 1:
            chi2, p, _, _ = chi2_contingency(tabela)
        else:
            p = np.nan
    except:
        p = np.nan

    return tabela, perc, p


# =========================
# VARIÁVEIS PARA ANÁLISE
# =========================

variaveis = {
    'FAIXA_IDADE': 'Faixa Etária',
    'TRATAM': 'Tratamento',
    'FORMA': 'Forma Clínica',
    'CS_SEXO': 'Sexo'
}

resultados = {}

print("\n========= RESULTADOS =========")

for var, nome in variaveis.items():

    if var == 'CS_SEXO':
        tabela, perc, p = analisar_associacao(df_filtrado, var)
    else:
        tabela, perc, p = analisar_associacao(df, var)

    resultados[var] = (tabela, perc, p)

    print(f"\n🔎 {nome}")
    print("Percentual de óbito (%):")
    print(perc.round(3))
    print(f"p-valor: {p}")


# =========================
# GRÁFICOS
# =========================

fig, axs = plt.subplots(2, 2, figsize=(16, 10))

for ax, (var, nome) in zip(axs.flat, variaveis.items()):

    tabela, perc, p = resultados[var]

    cores = cmap(np.linspace(0.3, 0.9, len(perc)))

    ax.bar(perc.index.astype(str), perc.values, color=cores)

    ax.set_title(f"{nome} vs Óbito")
    ax.set_ylabel("% de óbito")
    ax.tick_params(axis='x', rotation=25)

    # valores nas barras
    for i, v in enumerate(perc.values):
        ax.text(i, v + 0.02, f"{v:.2f}%", ha='center')

    # p-valor
    if not pd.isna(p):
        ax.text(
            0.5,
            max(perc.values)*0.85,
            f"p = {p:.4f}",
            ha='center'
        )

# =========================
# AJUSTES FINAIS
# =========================

for ax in axs.flat:
    ax.grid(alpha=0.3)

plt.subplots_adjust(hspace=0.35, wspace=0.25)
plt.show()
```

A análise indica que a **forma clínica da doença** é o principal fator associado ao óbito, com destaque para a forma hepatoesplênica, que apresenta taxa significativamente maior e forte evidência estatística.

```python
# Focos de esquistossomose por municipio (local provavel de infeccao)
try:
    import geobr
    import geopandas as gpd
except ImportError as exc:
    raise SystemExit("Instale geobr e geopandas para o mapa.") from exc

df_inf = df.copy()

# Mapa UF IBGE -> sigla (para lidar com COUFINF numerico)
mapa_uf = {
    11: "RO", 12: "AC", 13: "AM", 14: "RR", 15: "PA", 16: "AP", 17: "TO",
    21: "MA", 22: "PI", 23: "CE", 24: "RN", 25: "PB", 26: "PE", 27: "AL", 28: "SE", 29: "BA",
    31: "MG", 32: "ES", 33: "RJ", 35: "SP",
    41: "PR", 42: "SC", 43: "RS",
    50: "MS", 51: "MT", 52: "GO", 53: "DF",
}

# COUFINF pode estar como codigo (numerico) ou sigla (texto)
couf_num = pd.to_numeric(df_inf["COUFINF"], errors="coerce")
couf_sigla = df_inf["COUFINF"].astype("string").str.strip().str.upper()
df_inf["COUFINF_SIGLA"] = couf_num.map(mapa_uf)
df_inf.loc[df_inf["COUFINF_SIGLA"].isna(), "COUFINF_SIGLA"] = couf_sigla

df_inf["COMUNINF"] = df_inf["COMUNINF"].astype("string").str.strip().str.replace(".0", "", regex=False).str.zfill(6)

df_inf = df_inf[df_inf["COUFINF_SIGLA"].notna() & df_inf["COMUNINF"].notna()].copy()

def plot_focos_uf(uf_sigla: str, titulo: str) -> None:
    ag_uf = (
        df_inf[df_inf["COUFINF_SIGLA"] == uf_sigla]
        .groupby("COMUNINF")
        .size()
        .reset_index(name="casos")
    )
    ag_uf["COMUNINF"] = ag_uf["COMUNINF"].astype("string").str.zfill(6)

    if ag_uf.empty:
        print(f"Sem casos para {uf_sigla} com COUFINF/COMUNINF preenchidos.")
        return

    gdf_uf = geobr.read_municipality(code_muni="all", year=2020)
    gdf_uf = gdf_uf[gdf_uf["abbrev_state"] == uf_sigla]
    gdf_uf["code_muni_str"] = gdf_uf["code_muni"].astype("string").str[:6]

    gdf_plot = gdf_uf.merge(
        ag_uf,
        left_on="code_muni_str",
        right_on="COMUNINF",
        how="left",
    )
    gdf_plot["casos"] = gdf_plot["casos"].fillna(0)

    ax = gdf_plot.plot(
        column="casos",
        cmap="Reds",
        linewidth=0.1,
        figsize=(10, 10),
        legend=True,
        legend_kwds={"label": "Casos"},
    )
    ax.set_title(titulo)
    ax.set_axis_off()

# Ajuste os estados conforme necessario
plot_focos_uf("MG", "Focos por municipio - Minas Gerais (infecao provavel)")
plot_focos_uf("SE", "Focos por municipio - Sergipe (infecao provavel)")
```
