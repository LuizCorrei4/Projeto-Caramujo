```python
import pandas as pd
# ler um .parquet
df = pd.read_parquet('../data/raw/sinan_esqu_raw.parquet')
print(df.info())
print(df.describe())
```

```python
display(df.head(10))
```

```python
# contar quantos registros existem por atributo das colunas 'EVOLUCAO' e 'AN_QUALI'
print(df['EVOLUCAO'].value_counts())
print(df['AN_QUALI'].value_counts())

# contar quantos registros tem valores em branco ou nulos por atributo das colunas 'EVOLUCAO' e 'AN_QUALI'
print(df['EVOLUCAO'].isnull().sum())
print(df['AN_QUALI'].isnull().sum())

# preciso checar se os valores estão sem nada ou apenas com espaços em branco (que não são detectados por isnull())
print((df['EVOLUCAO'].str.strip() == '').sum())
print((df['AN_QUALI'].str.strip() == '').sum())


```
